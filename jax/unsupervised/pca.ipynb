# PCA in JAX

This notebook demonstrates unsupervised dimensionality reduction using Principal Component Analysis (PCA) implemented with JAX.

The example projects synthetic 3D data down to 2D and visualizes the result.

```python
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt

key = jax.random.PRNGKey(10)
data = jax.random.normal(key, (300, 3)) @ jnp.array([[1.0, 0.5, 0.2],[0.0,1.0,0.3],[0.0,0.0,0.7]]) + jnp.array([1.0, -1.0, 0.5])

mean = jnp.mean(data, axis=0)
centered = data - mean
cov = (centered.T @ centered) / (data.shape[0] - 1)
eigvals, eigvecs = jnp.linalg.eigh(cov)
order = jnp.argsort(eigvals)[::-1]
components = eigvecs[:, order[:2]]
projected = centered @ components
```

## Result

We visualize the two principal components to see how the high-dimensional structure projects into 2D.

```python
plt.figure(figsize=(7,6))
plt.scatter(projected[:,0], projected[:,1], s=20, alpha=0.8)
plt.title('PCA Projection (3D -> 2D)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```
