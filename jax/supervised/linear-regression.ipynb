# Linear Regression in JAX

This notebook demonstrates a supervised learning workflow using linear regression on synthetic data.

The example highlights parameter learning with gradient descent and mean squared error.

## Theory

Linear regression predicts a continuous target with a linear function:

$$
at{y} = wx + b
$$

The training objective is the mean squared error:

$$
athcal{L} = rac{1}{N} um_{i=1}^{N} (y_i - at{y}_i)^2
$$

```python
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt

key = jax.random.PRNGKey(0)
x = jnp.linspace(-3.0, 3.0, 120).reshape(-1, 1)
noise = 0.3 * jax.random.normal(key, (120, 1))
y = 4.0 * x - 1.5 + noise
```

```python
def predict(params, inputs):
    weight, bias = params
    return inputs * weight + bias

def loss_fn(params, inputs, targets):
    predictions = predict(params, inputs)
    return jnp.mean((predictions - targets) ** 2)

def train(inputs, targets, learning_rate=0.05, steps=300):
    params = (jnp.array([0.0]), jnp.array([0.0]))
    history = []
    grad_fn = jax.grad(loss_fn)
    for step in range(steps):
        gradients = grad_fn(params, inputs, targets)
        params = tuple(param - learning_rate * grad for param, grad in zip(params, gradients))
        if step % 50 == 0 or step == steps - 1:
            history.append((step, float(loss_fn(params, inputs, targets))))
    return params, history

params, history = train(x, y)
weight, bias = params
predictions = predict(params, x)
print(f'weight = {float(weight[0]):.3f}')
print(f'bias = {float(bias[0]):.3f}')
print(f'final loss = {float(loss_fn(params, x, y)):.4f}')
history
```

## Result

The learned line should closely match the synthetic data trend after optimization.

```python
plt.figure(figsize=(7, 5))
plt.scatter(x[:, 0], y[:, 0], s=25, alpha=0.7, label='data')
plt.plot(x[:, 0], predictions[:, 0], color='red', linewidth=2, label='fit')
plt.title('Linear Regression in JAX')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.legend()
plt.show()
```
